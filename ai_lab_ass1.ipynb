```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import tensorflow as tf
import cv2
from sklearn.preprocessing import LabelEncoder
print("="*60)
print("INTRODUCTION TO PYTHON LIBRARIES TO AI")
```

```python
print("\n----Numpy----")
numbers = np.array([10, 20, 30, 40, 50])
print("Array:", numbers)
print("Sum:", np.sum(numbers))
print("Max:", np.max(numbers))
print("Min:", np.min(numbers))
print("Median:", np.median(numbers))
```

```python
print("---PANDAS---")
Student = {
    "Name":["sayan","dipta","amrit"],
    "marks":[60,50,70]
}
df_student = pd.DataFrame(Student)
display(df_student)
```

```python
print("--Tensorflow--")
tensor1=tf.constant([1,2,3])
tensor2=tf.constant([4,5,6])
print("tensor1:",tensor1.numpy())
print("tensor2:",tensor2.numpy())
result=tf.add(tensor1,tensor2)
print("tensor addition:",result.numpy())
```

```python
print("\n----Matplotlib----")
print("Displaying bar chart..")
plt.figure(figsize=(4, 6))
plt.bar(df_student['Name'], df_student['marks'])
plt.xlabel('Student Name')
plt.ylabel('Marks')
plt.title('Student Marks Distribution')
plt.show()
```
